```python
# operasi elementwise (mengoperasikan setiap elemen yang berkorespondensi)
a = np.array([1, 2, 3, 4, 5])
b = np.array([6, 7, 8, 9, 10])

hasil1 = a + b
hasil2 = a - b
hasil3 = a * b
hasil4 = a / b
hasil5 = a ** 2

print("hasil 1 =", hasil1)
print("hasil 2 =", hasil2)
print("hasil 3 =", hasil3)
print("hasil 4 =", hasil4)
print("hasil 5 =", hasil5)
```

```python
# operator elementwise ada masalah di perkalian matriks
# karena mengalikan per elemen
a = np.array([(1,2,3),(4,5,6)])
b = np.array([(3,4,5),(3,4,5)])

hasil = a * b
print("hasil =", hasil)
```

```python
# perkalian matriks
a = np.array([(1,2,3),(4,5,6)])
b = np.array([(1,2,3),(1,2,3),(1,2,3)])

hasil_perkalian_matriks = np.dot(a,b)
print(hasil_perkalian_matriks)
```

```python

# mengetahui ukuran matriks
a = np.array([(1,2,3),(4,5,6)])
print("matriks a berukuran =", a.shape)

# transpose matriks
print("transpose matriks a =", a.transpose())
print("transpose matriks a =", np.transpose(a))
print("transpose matriks a =", a.T)

# pada Machine learning biasanya flatten dan stacking sangat penting
# flatten array, vektor baris
print("flatten dari matriks a =", a.ravel())
```

```python
# invers matriks
a = np.array([(1,-1),(1,1)])
print(a)

a_inv = np.linalg.inv(a)
print(a_inv)

det_a = np.linalg.det(a)
print(det_a)
```

```python
# solusi persamaan linear
A = np.array([(1,-6),(2,-5)])
Y = np.array([-35, 21])

print(A)
print(Y)

A_inv = np.linalg.inv(A)

# menyelesaikan persamaan linear
X1 = np.dot(A_inv, Y)
print(X1)

# cara lain
X2 = np.linalg.solve(A, Y)
print(X2)
```

```python
def f(x,y):
    return np.cos(np.exp(-x**2-y**2))*np.sin(np.sqrt(x**2+y**2))

x = np.linspace(-np.pi, np.pi, 30)
y = np.linspace(-np.pi, np.pi, 30)

X, Y = np.meshgrid(x, y)
Z = f(X, Y)

fig = plt.figure()
ax = plt.axes(projection = '3d')
ax.contour3D(X, Y, Z, 50, cmap = 'binary')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z');
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits import mplot3d

# ==========================================
# Nomor 1: Matriks 5x5 Random (Max, Min, Mean)
# ==========================================
# Membuat matriks 5x5 dengan nilai acak (0 sampai 1)
matriks = np.random.rand(5, 5)

nilai_max = np.max(matriks)
nilai_min = np.min(matriks)
nilai_mean = np.mean(matriks)

print("--- Nomor 1 ---")
print("Matriks 5x5:")
print(matriks)
print(f"Nilai Maksimum : {nilai_max}")
print(f"Nilai Minimum  : {nilai_min}")
print(f"Nilai Rata-rata: {nilai_mean}\n")


# ==========================================
# Nomor 2: Grafik 2 Dimensi (Implisit Contour)
# Persamaan: x^6 + 3x^4 y^2 - 3x^4 + 3x^2 y^4 - x^2 y^3 - 6x^2 y^2 + 3x^2 + y^6 - 3y^4 + 3y^2 = 1
# ==========================================
x = np.linspace(-2, 2, 400)
y = np.linspace(-2, 2, 400)
X, Y = np.meshgrid(x, y)

# Definisi persamaan f(x, y) = 1  =>  f(x, y) - 1 = 0
Z2 = (X**6 + 3*(X**4)*(Y**2) - 3*(X**4) + 3*(X**2)*(Y**4) 
      - (X**2)*(Y**3) - 6*(X**2)*(Y**2) + 3*(X**2) 
      + Y**6 - 3*(Y**4) + 3*(Y**2))

plt.figure(figsize=(6, 6))
# Menggambar garis kontur di mana nilai Z2 == 1
plt.contour(X, Y, Z2, levels=[1], colors='red')
plt.title("Grafik 2D: Persamaan Implisit")
plt.xlabel("x")
plt.ylabel("y")
plt.axhline(0, color='black', linewidth=0.5, linestyle='--')
plt.axvline(0, color='black', linewidth=0.5, linestyle='--')
plt.grid(True)
plt.show()


# ==========================================
# Nomor 3: Grafik 3 Dimensi
# Persamaan z = f(x, y) = ( (5y/4) - |x| )^2 + x^2
# ==========================================
x3 = np.linspace(-3, 3, 200)
y3 = np.linspace(-3, 3, 200)
X3, Y3 = np.meshgrid(x3, y3)

Z3 = ((5 * Y3 / 4) - np.abs(X3))**2 + X3**2

fig = plt.figure(figsize=(8, 6))
ax = plt.axes(projection='3d')

# Plot permukaan 3D
surf = ax.plot_surface(X3, Y3, Z3, cmap='viridis', edgecolor='none')
ax.set_title("Grafik 3D: z = ((5y/4) - |x|)^2 + x^2")
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
fig.colorbar(surf, ax=ax, shrink=0.5, aspect=5)
plt.show()
```
