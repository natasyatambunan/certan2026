```python
path = "weight-height_outlier.csv"
df2 = pd.read_csv(path)
df2
```

```python
import seaborn as sns
sns.boxplot(df2['Height'])
```

```python
import numpy as np

# Identifikasi Q1 Q3
q1, q3 = np.percentile(df2['Height'], [25, 75])

# Hitung IQR
iqr = q3 - q1

# Cari Lower dan upper bounds
lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)

# Hapus outliers
clean_data = df2[(df2['Height'] >= lower_bound) & (df2['Height'] <= upper_bound)]
sns.boxplot(clean_data['Height'])
```

```python
sns.boxplot(df2['Weight'])
```

```python
# Identifikasi Q1 Q3
q1, q3 = np.percentile(df2['Weight'], [25, 75])

# Hitung IQR
iqr = q3 - q1

# Cari Lower dan upper bounds
lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)

# Hapus outliers
clean_data = df2[(df2['Weight'] >= lower_bound) & (df2['Weight'] <= upper_bound)]
sns.boxplot(clean_data['Weight'])
```

```python
import matplotlib.pyplot as plt

# Membuat daftar kolom yang ingin dianalisis, yaitu "Weight" dan "Height".
columns_to_plot = ['Weight', 'Height']

# Memastikan kolom ada di dataset
columns_to_plot = [col for col in columns_to_plot if col in df2.columns]

fig, axs = plt.subplots(len(columns_to_plot), 1, dpi=95, figsize=(7, 5 * len(columns_to_plot)))

if len(columns_to_plot) == 1:
    axs = [axs]

for i, col in enumerate(columns_to_plot):
    axs[i].boxplot(df2[col], vert=False)
    axs[i].set_ylabel(col)

plt.show()
```

```python
def remove_outliers_sequentially(df2, columns):
    df_cleaned = df2.copy()
    for col in columns:
        q1, q3 = np.percentile(df_cleaned[col], [25, 75])
        iqr = q3 - q1
        lower_bound = q1 - (1.5 * iqr)
        upper_bound = q3 + (1.5 * iqr)
        df_cleaned = df_cleaned[(df_cleaned[col] >= lower_bound) & (df_cleaned[col] <= upper_bound)]
    return df_cleaned

# Terapkan fungsi untuk menghapus outliers dari "Height" dan "Weight"
clean_data = remove_outliers_sequentially(df2, ['Height', 'Weight'])

# Plot boxplots setelah menghapus outliers
columns_to_plot = ['Weight', 'Height']
fig, axs = plt.subplots(len(columns_to_plot), 1, dpi=95, figsize=(7, 5 * len(columns_to_plot)))

if len(columns_to_plot) == 1:
    axs = [axs]

for i, col in enumerate(columns_to_plot):
    axs[i].boxplot(clean_data[col], vert=False)
    axs[i].set_ylabel(col)

plt.show()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler(feature_range=(0, 1))

# Memisahkan kolom numerik dan non-numerik
numeric_cols = clean_data.select_dtypes(include=[np.number]).columns
non_numeric_cols = clean_data.select_dtypes(exclude=[np.number]).columns

rescaled_numeric = scaler.fit_transform(clean_data[numeric_cols])
rescaled_numeric
```

```python
# Konversi kembali hasil scaling ke DataFrame
rescaled_df = pd.DataFrame(rescaled_numeric, columns=numeric_cols, index=clean_data.index)

# Gabungkan kembali dengan kolom non-numerik tanpa mengubah jumlah baris
final_df = pd.concat([clean_data[non_numeric_cols], rescaled_df], axis=1)
final_df
```

```python
df.isnull().sum()
```

```python
df_filled = df.copy()

# Mengisi missing value pada Age dengan median
df_filled['Age'] = df_filled['Age'].fillna(df_filled['Age'].median())

# Mengisi missing value pada Cabin
df_filled['Cabin'] = df_filled['Cabin'].fillna('Unknown')

# Mengisi missing value pada Embarked dengan mode
df_filled['Embarked'] = df_filled['Embarked'].fillna(
    df_filled['Embarked'].mode().iloc[0]
)

df_filled
```

```python
df_filled.isnull().sum()
```

### penjelasan

Missing value pada dataset Titanic terdapat pada kolom Age, Cabin, dan Embarked. Missing value pada Age diisi menggunakan median, Cabin diisi dengan "Unknown", dan Embarked diisi menggunakan nilai mode. Setelah dilakukan proses tersebut, semua kolom sudah tidak memiliki missing value.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df2 = pd.read_csv("diabetes.csv")

df2
```

```python
columns = ['Glucose', 'BloodPressure', 'BMI']

for col in columns:
    plt.figure(figsize=(7, 4))
    plt.boxplot(df2[col])
    plt.title(f'Boxplot {col}')
    plt.ylabel(col)
    plt.show()
```

```python
for col in columns:
    q1 = df2[col].quantile(0.25)
    q3 = df2[col].quantile(0.75)
    
    iqr = q3 - q1
    
    lower_bound = q1 - (1.5 * iqr)
    upper_bound = q3 + (1.5 * iqr)
    
    print(f"\nKolom: {col}")
    print(f"Q1          : {q1}")
    print(f"Q3          : {q3}")
    print(f"IQR         : {iqr}")
    print(f"Lower Bound : {lower_bound}")
    print(f"Upper Bound : {upper_bound}")
```

```python
clean_data = df2.copy()

for col in columns:
    q1 = clean_data[col].quantile(0.25)
    q3 = clean_data[col].quantile(0.75)
    
    iqr = q3 - q1
    
    lower_bound = q1 - (1.5 * iqr)
    upper_bound = q3 + (1.5 * iqr)
    
    clean_data = clean_data[
        (clean_data[col] >= lower_bound) &
        (clean_data[col] <= upper_bound)
    ]

clean_data
```

```python
for col in columns:
    plt.figure(figsize=(7, 4))
    plt.boxplot(clean_data[col])
    plt.title(f'Boxplot {col} Setelah Handling Outlier')
    plt.ylabel(col)
    plt.show()
```

### penjelasan

Pada dataset diabetes.csv, deteksi outlier dilakukan pada tiga kolom, yaitu Glucose, BloodPressure, dan BMI. Deteksi dilakukan menggunakan boxplot dan metode IQR. Nilai yang berada di luar batas Q1 - 1.5 × IQR dan Q3 + 1.5 × IQR dianggap sebagai outlier dan dihapus. Setelah proses handling outlier, jumlah data berkurang dari 768 menjadi 709 baris.
