## TUGAS

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB

# 1. Load dataset Iris dan siapkan data
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]

X = df.drop(columns=["Species"])
y = df["Species"]

# 2. Encode label kategori menjadi angka
lab = LabelEncoder()
y = lab.fit_transform(y)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

model = GaussianNB()
NB_model = model.fit(X_train, y_train)

# PENGERJAAN TUGAS 2:
# Prediksi data baru: [7.3, 3.8, 4.6, 2.3]

data_baru = pd.DataFrame(
    np.array([[7.3, 3.8, 4.6, 2.3]]), 
    columns=iris.feature_names
)

hasil_prediksi = NB_model.predict(data_baru)

spesies_prediksi = lab.inverse_transform(hasil_prediksi)

print('Hasil prediksi untuk data baru [7.3, 3.8, 4.6, 2.3] adalah:', spesies_prediksi[0])
```

## Penjelasan
### diawali dengan mengemas nilai variabel data baru [7.3, 3.8, 4.6, 2.3] ke dalam bentuk DataFrame Pandas yang dilengkapi nama kolom sesuai dengan iris.feature_names (sepal length, sepal width, petal length, petal width). Penggunaan DataFrame ini bertujuan agar struktur data yang dimasukkan konsisten dengan format data yang digunakan saat melatih model, sehingga tidak memicu peringatan UserWarning pada library Scikit-Learn.   Setelah data siap, model GaussianNB memprediksi kelas data tersebut menggunakan fungsi NB_model.predict(data_baru). Hasil keluaran dari fungsi ini masih berupa nilai numerik terenkode, yaitu angka 2. Selanjutnya, nilai angka tersebut dikonversi kembali ke label kategori aslinya menggunakan fungsi lab.inverse_transform(hasil_prediksi). Dari langkah konversi ini, diperoleh hasil prediksi akhir berupa nama spesies bunga, yaitu virginica.
